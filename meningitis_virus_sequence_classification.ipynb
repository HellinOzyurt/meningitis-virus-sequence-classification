```python
from Bio import SeqIO

raw_klasor = os.path.join(ana_yol, 'raw_data')
clean_klasor = os.path.join(ana_yol, 'cleaned_data')

print("Başlıklar siliniyor ve saf dizilimler 'cleaned_data' klasörüne aktarılıyor...\n" + "-"*60)

# raw_data içindeki tüm dosyaları tara
for dosya in os.listdir(raw_klasor):
    if dosya.endswith(".fasta") or dosya.endswith(".fa"):
        girdi_yolu = os.path.join(raw_klasor, dosya)

        # Yeni dosya adı (Örn: CMV_raw_sequences.fasta -> CMV_cleaned.fasta)
        yeni_dosya_adi = dosya.replace("raw_sequences", "cleaned").replace(" ", "")
        cikti_yolu = os.path.join(clean_klasor, yeni_dosya_adi)

        # Dosyadaki tüm virüsleri listeye alalım
        virusler = list(SeqIO.parse(girdi_yolu, "fasta"))

        # Çıktı dosyasını "yazma" (w) modunda açıyoruz
        with open(cikti_yolu, "w") as f_out:
            for index, record in enumerate(virusler):
                # Sadece dizilimi (A,T,C,G harflerini) al ve metne çevir
                saf_dizilim = str(record.seq)

                # Dosyaya sadece saf dizilimi yaz
                f_out.write(saf_dizilim)

                # Eğer bu virüs, dosyadaki SON virüs değilse, araya İKİ satır atla
                # (Biri dizilimin bitişi için, diğeri senin istediğin 1 satır boşluk için)
                if index < len(virusler) - 1:
                    f_out.write("\n\n")

        print(f"✅ {dosya} işlendi: {len(virusler)} virüsün başlığı silindi ve aktarıldı.")

print("-" * 60 + "\nİŞLEM TAMAM! Tüm saf dizilimler 'cleaned_data' klasöründe hazır.")
```

```python
import os

# Sadece cleaned_data (2. aşama) klasörümüzü okuyacağız
ana_yol = '/content/drive/MyDrive/thesis/viruses/'
clean_klasor = os.path.join(ana_yol, 'cleaned_data')

print("🔍 VERİLERİN RÖNTGENİ ÇEKİLİYOR ('N' Analiz Raporu)...\n" + "="*60)

for dosya in os.listdir(clean_klasor):
    if dosya.endswith(".fasta") or dosya.endswith(".fa"):
        girdi_yolu = os.path.join(clean_klasor, dosya)

        with open(girdi_yolu, 'r') as f_in:
            icerik = f_in.read()

        # Virüsleri boşluklardan ayır
        dizilimler = [d.strip() for d in icerik.split('\n\n') if d.strip()]

        toplam = len(dizilimler)
        n_yok = 0
        n_kabul_edilebilir = 0
        n_cop = 0

        print(f"\n📁 DOSYA: {dosya} (Toplam {toplam} Virüs)")
        print("-" * 40)

        for index, seq in enumerate(dizilimler):
            n_sayisi = seq.upper().count('N')
            n_orani = n_sayisi / len(seq) if len(seq) > 0 else 0

            if n_sayisi == 0:
                n_yok += 1
            elif n_orani <= 0.01:
                n_kabul_edilebilir += 1
            else:
                n_cop += 1

        # Raporu Yazdır
        print(f"   🟢 Kusursuz (İçinde hiç 'N' olmayan): {n_yok} adet")
        print(f"   🟡 Kabul Edilebilir (İçinde çok az 'N' olan, <= %1): {n_kabul_edilebilir} adet")
        print(f"   🔴 ÇÖPE ATILACAK (Sınırı aşan bozuk veriler, > %1): {n_cop} adet")
        print(f"   ➡️ Filtreleme yaparsak YENİ DOSYADA KALACAK TOPLAM VİRÜS: {n_yok + n_kabul_edilebilir}")

print("\n" + "="*60)
print("RAPOR TAMAMLANDI. (Henüz hiçbir veri silinmedi veya taşınmadı).")
```

```python
import os

# Klasör Yolları
ana_yol = '/content/drive/MyDrive/thesis/viruses/'
clean_klasor = os.path.join(ana_yol, 'cleaned_data')
filtered_klasor = os.path.join(ana_yol, 'filtered_data')

# N harfi için sınırımız (%1)
max_n_orani = 0.01

print("'cleaned_data' klasörü okunuyor ve N-Filtrelemesi başlıyor...\n" + "-"*65)

# cleaned_data içindeki dosyaları tek tek tara
for dosya in os.listdir(clean_klasor):
    if dosya.endswith(".fasta") or dosya.endswith(".fa"):
        girdi_yolu = os.path.join(clean_klasor, dosya)

        # Yeni dosya adı (Örn: CMV_cleaned.fasta -> CMV_filtered.fasta)
        yeni_dosya_adi = dosya.replace("cleaned", "filtered")
        cikti_yolu = os.path.join(filtered_klasor, yeni_dosya_adi)

        # Dosyayı okuyup içindeki metni alıyoruz
        with open(girdi_yolu, 'r') as f_in:
            icerik = f_in.read()

        # Hatırlarsan virüslerin arasına iki satır boşluk (\n\n) koymuştuk, oradan bölüyoruz
        dizilimler = icerik.split('\n\n')

        temiz_kalanlar = []

        # Her bir virüs dizilimini tek tek incele
        for dizilim in dizilimler:
            dizilim = dizilim.strip().upper() # Etrafındaki görünmez boşlukları sil ve büyük harf yap

            if len(dizilim) > 0: # Boş bir satır gelmediğinden emin ol
                n_sayisi = dizilim.count('N')
                n_orani = n_sayisi / len(dizilim)

                # N oranı %1'den küçükse (veya eşitse), temiz listesine al
                if n_orani <= max_n_orani:
                    temiz_kalanlar.append(dizilim)

        # Temiz kalanları yepyeni filtered_data klasörüne aynı formatta yaz
        with open(cikti_yolu, 'w') as f_out:
            for index, seq in enumerate(temiz_kalanlar):
                f_out.write(seq)
                if index < len(temiz_kalanlar) - 1:
                    f_out.write("\n\n") # Yine aralarına boşluk koyduk

        atilan_virus = len(dizilimler) - len(temiz_kalanlar)
        print(f"✅ {dosya}: {len(dizilimler)} virüs incelendi. -> {atilan_virus} tanesi ('N' yüzünden) çöpe atıldı, {len(temiz_kalanlar)} tanesi kurtarıldı.")

print("-" * 65 + "\nİŞLEM TAMAM! N-Filtresinden geçen en kaliteli veriler 'filtered_data' klasörüne aktarıldı.")
```

```python
import os

# 1. Klasör Yolları
ana_yol = '/content/drive/MyDrive/thesis/viruses/'
clean_klasor = os.path.join(ana_yol, 'cleaned_data')
tested_klasor = os.path.join(ana_yol, 'tested') # Hocanın istediği test klasörü

# Sadece klasörü oluşturuyoruz
if not os.path.exists(tested_klasor):
    os.makedirs(tested_klasor)
    print(f"📁 Yeni klasör başarıyla oluşturuldu: {tested_klasor}\n")
else:
    print(f"📁 Klasör zaten mevcut: {tested_klasor}\n")

# 2. Esma Hocanın önerdiği yeni sınır: %5 (0.05)
max_n_orani = 0.02

print("Esma Hocanın talimatlarına göre %5 'N' Filtrelemesi başlıyor...\n" + "-"*65)

# 3. İşlem ve Filtreleme
for dosya in os.listdir(clean_klasor):
    if dosya.endswith(".txt") or dosya.endswith(".fasta") or dosya.endswith(".fa"):
        girdi_yolu = os.path.join(clean_klasor, dosya)

        # Yeni dosya adı (Örn: CMV_cleaned.txt -> CMV_tested.txt)
        yeni_dosya_adi = dosya.replace("cleaned", "tested")
        cikti_yolu = os.path.join(tested_klasor, yeni_dosya_adi)

        # Dosyayı oku
        with open(girdi_yolu, 'r') as f_in:
            icerik = f_in.read()

        # Virüsleri iki satır boşluğundan ayır
        dizilimler = [d.strip() for d in icerik.split('\n\n') if d.strip()]

        temiz_kalanlar = []

        # Her bir dizilimi %5 kuralına göre test et
        for seq in dizilimler:
            n_sayisi = seq.upper().count('N')
            n_orani = n_sayisi / len(seq) if len(seq) > 0 else 0

            # N oranı %5'ten küçükse veya eşitse al (İçindeki N'lere HİÇ DOKUNMADAN)
            if n_orani <= max_n_orani:
                temiz_kalanlar.append(seq)

        # Testi geçenleri yeni "tested" klasörüne yaz
        with open(cikti_yolu, 'w') as f_out:
            for index, seq in enumerate(temiz_kalanlar):
                f_out.write(seq)
                if index < len(temiz_kalanlar) - 1:
                    f_out.write("\n\n") # Aralarına yine 1 satır boşluk koyuyoruz

        atilan_virus = len(dizilimler) - len(temiz_kalanlar)
        print(f"✅ {dosya}: {len(dizilimler)} virüs incelendi. -> {atilan_virus} tanesi elendi, {len(temiz_kalanlar)} tanesi (%5 kuralıyla) kurtarıldı.")

print("-" * 65 + "\nİŞLEM TAMAM! %5 sınırını geçen veriler hiçbir değişikliğe uğramadan 'tested' klasörüne aktarıldı.")
```
